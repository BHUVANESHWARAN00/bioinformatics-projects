```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
df =  pd.read_csv("../data/raw/luad_tcga_gdc_clinical_data.tsv", sep="\t")
```

```python
columns = [
    "Patient ID",
    "Diagnosis Age",
    "Sex",
    "Race Category",
    "Ethnicity Category",
    "Years Smoked",
    "Person Cigarette Smoking History Pack Year Value",
    "AJCC Pathologic Stage",
    "AJCC Pathologic T-Stage",
    "AJCC Pathologic N-Stage",
    "AJCC Pathologic M-Stage",
    "Primary Diagnosis",
    "Overall Survival (Months)",
    "Overall Survival Status",
    "Disease Free (Months)",
    "Disease Free Status",
    "Mutation Count",
    "Fraction Genome Altered",
    "Patient's Vital Status",
    "Year of Diagnosis"
]

clinical_df = df.loc[:, columns].copy()

print("Shape:", clinical_df.shape)


clinical_df.head()
```

```python
clinical_df = clinical_df.drop_duplicates(subset="Patient ID")
print("Shape after removing duplicates:", clinical_df.shape)        
```

```python
clinical_df["Overall Survival Status"] = clinical_df["Overall Survival Status"].replace({"0:LIVING":"Living", "1:DECEASED":"Deceased"})   
```

```python
clinical_df["Patient's Vital Status"] = clinical_df["Patient's Vital Status"].replace({
    "Alive": "Alive",
    "Dead": "Dead"
})
```

```python
clinical_df["Age Group"] = pd.cut(
    clinical_df["Diagnosis Age"],
    bins=[0,50,60,70,80,100],
    labels=["<50","50-60","60-70","70-80","80+"]
)
```

```python
clinical_df["Race Category"] = clinical_df["Race Category"].fillna("Unknown")
clinical_df["Ethnicity Category"] = clinical_df["Ethnicity Category"].fillna("Unknown")
clinical_df["Years Smoked"] = clinical_df["Years Smoked"].fillna(0)
clinical_df["Person Cigarette Smoking History Pack Year Value"] = clinical_df["Person Cigarette Smoking History Pack Year Value"].fillna(0)
```

```python
clinical_df.info()
```

```python
clinical_df.to_csv(
    "../data/processed/luad_clinical_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
