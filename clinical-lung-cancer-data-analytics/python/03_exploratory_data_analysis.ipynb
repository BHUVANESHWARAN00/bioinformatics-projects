```python
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)

df = pd.read_csv("../data/processed/luad_clinical_cleaned.csv")
```

```python
df.shape
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    df["Diagnosis Age"].dropna(),
    bins=15,
    edgecolor="black",
    linewidth=1
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")

plt.tight_layout()
plt.savefig(
    "../images/eda/age_distribution.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```

```python
plt.figure(figsize=(6,4))

df["Sex"].value_counts().plot(
    kind="bar",
    edgecolor="black",
)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")

plt.grid(axis="y", linestyle="--", alpha=0.4)

ax = plt.gca()
for spine in ax.spines.values():
    spine.set_edgecolor("black")
    spine.set_linewidth(1.5)

plt.tight_layout()
plt.savefig(
    "../images/eda/gender_distribution.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```

```python
plt.figure(figsize=(6,6))

df["Patient's Vital Status"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    wedgeprops={"edgecolor":"black","linewidth":1},
)

plt.title("Patient Vital Status")
plt.ylabel("")

ax = plt.gca()
for spine in ax.spines.values():
    spine.set_edgecolor("black")
    spine.set_linewidth(1.5)

plt.tight_layout()
plt.savefig(
    "../images/eda/vital_status.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Mutation Count"].dropna(),
         bins=20,
         edgecolor="black")

plt.title("Mutation Count Distribution")
plt.xlabel("Mutation Count")
plt.ylabel("Number of Patients")

plt.grid(axis="y", linestyle="--", alpha=0.4)

ax = plt.gca()
for spine in ax.spines.values():
    spine.set_edgecolor("black")
    spine.set_linewidth(1.5)

plt.tight_layout()
plt.savefig(
    "../images/eda/mutation_count_distribution.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```

```python
plt.figure(figsize=(8,5))

df["Age Group"].value_counts().sort_index().plot(
    kind="bar",
    edgecolor="black",
)

plt.title("Age Group Distribution")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")

plt.grid(axis="y", linestyle="--", alpha=0.4)

ax = plt.gca()
for spine in ax.spines.values():
    spine.set_edgecolor("black")
    spine.set_linewidth(1.5)

plt.tight_layout()
plt.savefig(
    "../images/eda/age_group_distribution.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```

```python
plt.figure(figsize=(8,5))

df["Race Category"].value_counts().plot(
    kind="bar",
    edgecolor="black",
)

plt.title("Race Distribution")
plt.xlabel("Race")
plt.ylabel("Number of Patients")

plt.xticks(rotation=45)

plt.grid(axis="y", linestyle="--", alpha=0.4)

ax = plt.gca()
for spine in ax.spines.values():
    spine.set_edgecolor("black")
    spine.set_linewidth(1.5)

plt.tight_layout()
plt.savefig(
    "../images/eda/race_distribution.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```

```python
import matplotlib.pyplot as plt

numeric_df = df.select_dtypes(include=["float64", "int64"])

corr = numeric_df.corr()

plt.figure(figsize=(10,8))

plt.imshow(corr, cmap="coolwarm", aspect="auto")

plt.colorbar()

plt.xticks(range(len(corr.columns)),
           corr.columns,
           rotation=90)

plt.yticks(range(len(corr.columns)),
           corr.columns)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.savefig(
    "../images/eda/correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```

```python
stage_vital = pd.crosstab(
    df["AJCC Pathologic Stage"],
    df["Patient's Vital Status"],
)

stage_vital.plot(
    kind="bar",
    figsize=(10,6),
    edgecolor="black",
)

plt.title("Cancer Stage vs Vital Status")
plt.xlabel("Cancer Stage")
plt.ylabel("Number of Patients")

plt.grid(axis="y", linestyle="--", alpha=0.4)

ax = plt.gca()
for spine in ax.spines.values():
    spine.set_edgecolor("black")
    spine.set_linewidth(1.5)

plt.tight_layout()
plt.savefig(
    "../images/eda/stage_vital.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```

```python
age_vital = pd.crosstab(
    df["Age Group"],
    df["Patient's Vital Status"],
)

age_vital.plot(
    kind="bar",
    figsize=(8,5),
    edgecolor="black",
)

plt.title("Age Group vs Vital Status")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")

plt.grid(axis="y", linestyle="--", alpha=0.4)

ax = plt.gca()
for spine in ax.spines.values():
    spine.set_edgecolor("black")
    spine.set_linewidth(1.5)

plt.tight_layout()
plt.savefig(
    "../images/eda/age_group_vs_vital_status.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Years Smoked"],
    df["Overall Survival (Months)"],
    alpha=0.6,
)

plt.title("Smoking Years vs Overall Survival")
plt.xlabel("Years Smoked")
plt.ylabel("Overall Survival (Months)")

plt.grid(alpha=0.4)

ax = plt.gca()
for spine in ax.spines.values():
    spine.set_edgecolor("black")
    spine.set_linewidth(1.5)

plt.tight_layout()
plt.savefig(
    "../images/eda/years_smoked_vs_survival.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```
